```python
import scanpy as sc
import numpy as np
import scipy.sparse as sp

def mask_adata(adata, mask_fraction=0.1, seed=None):

    if seed is not None:
        np.random.seed(seed)
    X_orig = adata.X

    if sp.issparse(X_orig):
        X = X_orig.toarray()
    else:
        X = X_orig.copy()

    mask = np.random.rand(*X.shape) < mask_fraction
    X[mask] = 0

    adata_masked = adata.copy()
    if sp.issparse(X_orig):
        adata_masked.X = sp.csr_matrix(X)
    else:
        adata_masked.X = X
    return adata_masked

adata1 = sc.read_h5ad("00_data/12_Slide_tags/adata_RNA.h5ad")
adata2 = sc.read_h5ad("/00_data/12_Slide_tags/adata_ATAC.h5ad")

# mask_fraction = [0.1, 0.3, 0.5, 0.7, 0.9]
adata1_masked = mask_adata(adata1, mask_fraction=0.9, seed=42)
adata2_masked = mask_adata(adata2, mask_fraction=0.9, seed=42)

print(f"Zero-value ratio of RNA before masking: {np.mean(adata1.X == 0):.2%}")
print(f"Zero-value ratio of RNA after masking: {np.mean(adata1_masked.X == 0):.2%}")

print(f"Zero-value ratio of ATAC before masking: {np.mean(adata2.X == 0):.2%}")
print(f"Zero-value ratio of ATAC after masking: {np.mean(adata2_masked.X == 0):.2%}")

nan_data1 = np.isnan(adata1_masked.X).sum()
nan_data2 = np.isnan(adata2_masked.X).sum()
print(f"NaN values in data1: {nan_data1}")
print(f"NaN values in data2: {nan_data2}")
print(np.isinf(adata1_masked.X).sum())
print(np.isnan(adata2_masked.X).sum())

# adata1_masked.write("/00_data/14_add_noise_and_mask_dataset/12_Slide_tags_mask_09/adata_RNA.h5ad")
# adata2_masked.write("/00_data/14_add_noise_and_mask_dataset/12_Slide_tags_mask_09/adata_ATAC.h5ad")
```

```python
import scanpy as sc
import numpy as np
import scipy.sparse as sp

def mask_adata(adata, mask_fraction=0.1, seed=None):
    if seed is not None:
        np.random.seed(seed)
    X_orig = adata.X

    if sp.issparse(X_orig):
        X = X_orig.toarray()
    else:
        X = X_orig.copy()
        
    mask = (np.random.rand(*X.shape) < mask_fraction) & (X != 0)
    X[mask] = 0

    adata_masked = adata.copy()
    if sp.issparse(X_orig):
        adata_masked.X = sp.csr_matrix(X)
    else:
        adata_masked.X = X

    return adata_masked

adata1 = sc.read_h5ad("/00_data/02_Human_tonsil_slice1/adata_RNA.h5ad")
adata2 = sc.read_h5ad("/00_data/02_Human_tonsil_slice1/adata_ADT.h5ad")

adata1_masked = mask_adata(adata1, mask_fraction=0.9, seed=42)
adata2_masked = mask_adata(adata2, mask_fraction=0.9, seed=42)

print(f"Zero-value ratio of RNA before masking: {np.mean(adata1.X == 0):.2%}")
print(f"Zero-value ratio of RNA after masking: {np.mean(adata1_masked.X == 0):.2%}")

print(f"Zero-value ratio of ADT before masking: {np.mean(adata2.X == 0):.2%}")
print(f"Zero-value ratio of ADT after masking: {np.mean(adata2_masked.X == 0):.2%}")

# adata1_masked.write("/00_data/14_add_noise_and_mask_dataset/02_Human_tonsil_slice1_mask_09/adata_RNA.h5ad")
# adata2_masked.write("/00_data/14_add_noise_and_mask_dataset/02_Human_tonsil_slice1_mask_09/adata_ADT.h5ad")

```
